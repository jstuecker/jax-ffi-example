```python
import jax
import jax.numpy as jnp
from jax_ffi_example import forces
```

```python
a = jax.random.uniform(jax.random.key(0), 100)
b = jax.random.uniform(jax.random.key(1), 100)

c = forces.multiply(a, b)

print(jnp.allclose(c, a*b))
```

```python
pos = jax.random.normal(jax.random.key(0), (10000,3))
mass = jnp.ones_like(pos[...,0]) 

force = forces.cuda_force(pos, mass, softening=0.1)
force2 = forces.jax_force(pos, mass, softening=0.1)
force3 = forces.simple_cuda_force(pos, mass, softening=0.1, gap=0)

print(jnp.allclose(force, force2, atol=1e-2))
print(jnp.allclose(force, force3, atol=1e-2))
```
